### Recursive Function
Recursive function we can say that a function is going to call itself again and again 

```python
## Factorial of a Number
def factorial(num):
    if num == 0 or num == 1:
        return 1
    elif num < 0:
        print("Factorial is not defined for negative numbers.")
    else:
        # fact = 1
        return num * factorial(num-1)


n = int(input("Enter the number:"))
print(f"factorial of {n} is :- {factorial(n)}")
```

#### Positional Argument
A positional argument is an argument passed to a function where the position (order) of the argument determines which parameter receives the value.

Example

```python
def student(name, age):
    print(f"Name: {name}")
    print(f"Age: {age}")

student("Vinod", 27)
```

```python
### Variable Length Arguments
## Positional And Keywords arguments

def print_numbers(*krish):
    for number in krish:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
### Keywords Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Krish",age="32",country="India")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional arument :{val}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Krish",name="Krish",age="32",country="India")
```

### Positional Only Arguments

To specify positional-only arguments, add , / after the arguments:

```python
def my_function(name, /):
  print("Hello", name)

my_function("Emil")
```

Without the , / you are actually allowed to use keyword arguments even if the function expects positional arguments:

```python
def my_function(name):
  print("Hello", name)

my_function(name = "Emil")
```

```python
## 1. To create a function that converts temperature from celcius to Fahrenheit or Fahrenheit to celcius
def temperature_change(temp, unit):
    if unit == 'C':
        temperature=temp * (9/5) + 32
        print(f"the temperature for {temp} celcius is {temperature} Fahrenheit")
    elif unit == "F":
        temperature= (temp - 32) * 5/9
        print(f"the temperature for {temp} Fahrenheit is {temperature} celcius  ")
    else:
        print("Invalid")

temperature_change(1, 'C')
```

```python
## Define a Function that would behave as a Password Strenght checker
def password_strength(password):
    if len(password) < 8:
        print(f"The entered password {password} is a weak password")
    elif not any (char.islower() for char in password):
        print(f"The entered password {password} is a Weak Password")
    elif not any (char.isupper() for char in password):
        print(f"The entered password {password} is a Weak Password")
    elif not any(char.isdigit() for char in password):
        print(f"The entered password {password} is a Weak Password")
    elif not any (char in "!@$#%^*()-=_+" for char in password):
        print(f"The entered password {password} is a Weak Password")
    elif len(password) > 8 and any(char.isupper() for char in password) and any (char.islower() for char in password) and any(char.isdigit() for char in password) and any (char in "!@$#%^*()-=_+" for char in password): 
        print(f"The entered password {password} is a Strong Password")

password_strength("HelloVinod123#$")
```

```python
## 2. Check that wheather if the currect String is a palindrom or not
def palindrom(string):
    string_value=string.lower().replace(" ", "")

    if string_value==string_value[::-1]:
        print("The current string is a palindrom")
    else:
        print("The current string is not a palindrom")

palindrom('Ada')
palindrom('ad')
```

```python
## Calculate the total cost of items in a shopping cart
cart = [
        {'item1':'Apple','Cost':5, 'Quantity':20},
        {'item1':'Banana','Cost':15, 'Quantity':10},
        {'item1':'Guava','Cost':51, 'Quantity':15},
    ]

def cart_total(cart):
    Total_cost=0
    for item in cart:
        Total_cost+=item['Cost']*item['Quantity']
    print("The Total cost for the Cart is :", Total_cost)

cart_total(cart)
```
