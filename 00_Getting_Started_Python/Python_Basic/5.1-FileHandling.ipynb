### File Handling

#### 1. Read The content forom my text file in the system

- Open() --> Opening your file
- First parameter --> File name
- Second Parameter ---> Mode
- Mode --> It deside the way in which we are working with this file
- `Mode --> r` --> Reading the content from the file
- `Mode --> w` --> Write the content into the File
- `Mode --> r+` --> Read the content First and then writing the content into the File
- `Mode --> w+` --> Writing the content first and then reading from it

```python
## Reading a file
f=open('example.txt', 'r')
f.read()
```

```python
## Write() --> helps us out in writing the content into the file

## wirte() --> it writes the content into the text file
    # Mode --> W --> Content will be overwritten
    # Mode --> a --> Content will be added into our file
## append() --> it adds the data from the end of the file
```

```python
f=open('example.txt', 'a')
f.write('\n We are learning Advance Backend')
f.close()
```

#### with --> It will be used for first of all opening the file as a object and there is norequirement for closing the file as well

```python
with open('example.txt', 'r') as f:
    content_data=f.read()
    print(content_data)
```

```python
with open('example.txt', 'w') as file:
    file.write("we are exploring about with command in python file handling")
```

```python
with open('example.txt', 'a') as file:
    file.write("\n Hello we are exploring append mode")
```

### Write the content into our file line by line --> writelines()

```python
lines = ['First_line\n', 'Second_line\n', 'Third_line']
with open('example.txt', 'w') as data:
    data.writelines(lines)
```

### Mode --> Write the content into the binary file --> wb
### Mode --> Read the content from the binary file --> rb

```python
data=b'\x00\x01\x02\x03\x04'
with open('example.txt', 'wb') as binary_data:
    binary_data.write(data)
```

```python
with open('example.txt', 'rb') as binary_data:
    binary_content=binary_data.read()
    print(binary_content)
```

### Read the content from the source.txt file and writting the content inot another file


```python
with open('source.txt', 'r') as source:
    content=source.read()

with open('destination.txt', 'w') as destination:
    destination.write(content)

```

### Reading the Total Lines, Total words, Total Characters from a file --> readlines()

```python
def count_line(file):
    with open(file, 'r') as files:
        lines=files.readlines()
        total_lines=len(lines)
        total_words= sum(len(line.split()) for line in lines)
        total_characters=sum(len(line) for line in lines)
    return total_lines, total_words, total_characters


count_line('Total_data.txt')
```

### Write the content into a file and then after writing the content into that file we are going to read the content from that file as well --> seek()

W+ ---> First of all as W is their so it states that first writing operation will be done and then Reading Operation will be done

Seek() --> it is going to help us out  with reading the content after writing from the starting only

```python
with open('Total_data.txt', 'w+') as write_read_fiel:
    write_read_fiel.write("\n We are looking into W+ operation")
    write_read_fiel.write("\n It mean writing and then Reading")
    write_read_fiel.seek(0)
    content_complete_data=write_read_fiel.read()
    print(content_complete_data)
```
