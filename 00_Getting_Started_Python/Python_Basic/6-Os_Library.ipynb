## OS Libary

OS Library is able to inreract with our Operating System and can let us work with Multiple Directories and Files

```python
## getcwd() --> It is capable of providing us with the idea about the current working directory
## getcwd() --> get current working directory
```

```python
import os
```

```python
current_path=os.getcwd()
print("The current working directory is:", current_path)
```

```python
## mkdir() --> make directory --> It will be helping us out with creating the directory
```

```python
new_path="New_Folder"
os.mkdir(new_path)
print(f"The new directory {new_path} is created successfully")
```

```python
## listdir() --> Listing all of the directory present
All_Items=os.listdir('.')
print(All_Items)
```

```python
## Joining multiple paths Together
# os.path.join() --> It will be joining multiple paths together
dir_name="Folder_data"
file_name="File_Data"
Full_Path=os.path.join(dir_name, file_name)
print(f"The Full path is: {Full_Path}")
```

```python
## I Want to join multiple path with the current path
current_complete_path=os.path.join(os.getcwd(), dir_name, file_name)
print(f"The complete path is : {current_complete_path}")
```

```python
## os.path.exists()
 
given_path="d:\GitHub_Repo\dsa-in-python\00_Getting_Started_Python\Python_Basic"
if os.path.exists(given_path):
    print("Yes this path exists")
else:
    print("This path does not exists")
```

```python
## os.path.isfile() --> This checks that wheter if the current path is a file or not
## os.path.isdir() --> This checks tht whether if the current path is a directory or not

path='Total_data.txt'

if os.path.isfile(path):
    print("The current path belongs to a file")

elif os.path.isdir(path):
    print("The current path belongs to a directory")

else:
    print("The current path is neither a file or a folder")
```

```python
## os.path.abspath()--> It provides us with the complete path of the absolute path of the file or directory
file_name='Total_data.txt'
complete_path=os.path.abspath(file_name)

print("The complete path or absolute path for the given file is :", complete_path)
```
