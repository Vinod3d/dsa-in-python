```python
## csv library --> Most of the time wa are going to work with csv  files so thats why knowledge with csv library functionality matters
```

```python
import csv
```

```python
with open('example.txt', 'w', newline="") as file:
    writer=csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Amit', 'age'])
    
```

```python
with open('example.txt', 'r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Datetime Library --> it helps us out with figuring out the current time and date
```

```python
from datetime import datetime,timedelta
```

```python
time_current=datetime.now()
print(time_current)
```
