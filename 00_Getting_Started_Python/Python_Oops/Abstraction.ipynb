## Abstraction

Abstraction is the concept of Hinding the implementation details and only showing the necessary features of the object.

This helps in reducing programming complexity and efforts

Abstraction allow us to work with an object with what an object does rather than how it does it.

### How can we achieve abstraction in python

#### 1. Abstract class
A Class that cannot be instatiated or that means for an abstract class we cant create the object of this class.

#### 2. Abstract method
An abstract method is a method that is declared in the abstract class but whose implementation is provided by the child class.

The abstract method is being defined by @abstractmethod

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("The car start with an ignition key")

    def stop(self):
        print("The car stops with break pedals")


class ElectricScooter(Vehicle):
    def start(self):
        print("Electric Scooter is going to start with a power button")

    def stop(self):
        print("Electric Scooter stops with an electric break")

```

```python
car = Car()
```

```python
scooter = ElectricScooter()
```

```python
scooter.start()
```
