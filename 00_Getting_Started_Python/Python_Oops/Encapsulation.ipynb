### Encapsulation
Encapsulation is the process of bundling data and the methods that operate on that data into a single unit, while restricting direct access to the internal data.

#### Key concepts of Encapsulation


#### 1. Public Member
- Accessible from Anywhere or Attributes are accessible outside of the class and these members are accessible by the derived class 
- public member attribute is directly declared.

```python
class Ironman:
    def __init__(self, name, power, weapon):
        self.name = name
        self.power = power
        self.weapon = weapon


    def introduction(self):
        print(f"Hello my name is {self.name} and i have {self.weapon}")
```

**dir():** It takes the input as the object name and gives us this idea that which member is public 

```python
iron=Ironman("Amit", "Genuis Intellect", "Repulsor Beams")
```

```python
dir(iron)
```

```python
iron.introduction()
```

#### 2. Private Member
A Private member is not accessible outside the class and not even in the derived class and in order to have the access over these private members or to bring some modification on these members we use getter and setter method.

Private member are declared by using double underscore __ before the name

```python
class Avenger:
    def __init__(self, name, phone_number):
        self.name=name
        self.__phone_number=phone_number
        
```

```python
avenger = Avenger("captain", "7876846847")
```

```python
avenger.name
```

```python
avenger.phone_number
```

```python
dir(avenger)
```

#### Getter function
It will be helping us out with accessing the private data

```python
class Captain:
    def __init__(self, name, address):
        self.name=name
        self.__address = address

    def accessing_add(self):
        print(f"My address is {self.__address}")
```

```python
captain = Captain("Mohit", "New Delhi")
```

```python
captain.accessing_add()
```

#### Setter Method
Setter Method is being used for modifying our private data

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.__sound = sound

    def sound_info(self):
        print(f"The Animal is going to sound as :{self.__sound}")

    def sound_modify(self, new_sound):
        self.__sound= new_sound
        print(f"New sound for the Animal is : {self.__sound}")
```

```python
animal = Animal("Bruno", "Woof Woof")
```

```python
animal.name
```

```python
animal.sound_info()
```

```python
animal.sound_modify("meaw meaw")
```

#### 3. Protected Member

Protected Data is the data which is not accessible outside the class but thre protected data is going to be accessible by the derived class or child class

For having the protected member you can mention your protected member as _attribute or  by using a single underscore

```python
class GrandParent:
    def __init__(self, name, locker_key):
        self.name=name
        self._locker_key=locker_key

    def locker_info(self):
        print(f"The locker key info is {self._locker_key}")
```

```python
class Parent(GrandParent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)

    def child_reveals(self):
        print(f"The locker key information is: {self._locker_key}")
```

```python
grandpa=GrandParent("Amit", "1244454")
```

```python
grandpa.name
```

```python
grandpa._locker_key
```

```python
dir(grandpa)
```

```python
parent = Parent("Amit", "134534")
```

```python
parent._locker_key
```

```python
parent.child_reveals()
```

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder=account_holder
        self.__balance=balance

    def get_balance(self):
        print(f"The Total balance is: {self.__balance}")

    def deposit(self, amount):
        if amount>0:
            self.__balance+=amount
            return f"Deposit Amount is: {self.get_balance}"
        else:
            return "Deposited Amount must be positive"

    def withdraw(self, amount):
        if 0<amount<=self.__balance:
            self.__balance-=amount
            return f"Withdrawal {amount} is : {self.get_balance()}"
        else:
            return "Insufficient funds in the account"
```

```python
account = BankAccount("Vinod", 5000)
```

```python
account.get_balance()
```

```python
account.withdraw(200)
```
