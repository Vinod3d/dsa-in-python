# Pillars of Object Oriented Programming

## Inheritance

Inheritance is a fundamental concept in Object Oriented programming because of which the child class is able to inherit the properties OR Attributes and functionalities of the parent class.

Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definitiely in this hierarchy the grandchild can have some features different from the parent and grandparent as well.

### In Python typically we get to see 3 Types of Inheritance :

#### 1. Single Inheritance:
The Features or the Attributes from one parent class will be getting inherited into one child class.



```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power= power

    def introduction(self):
        print(f"Hello My name is {self.name}")


class Ironman(Avenger):
    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon=weapon

    def weapon_details(self):
        print(f"I use {self.weapon}")
    

iron = Ironman("Ironman", "Genius Intellect", "Repulsor beams")
iron.name
iron.weapon
iron.weapon_details()
```

### 2. Multilevel Inheritance
The properties or Methods and Attributes of the grandparent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class. In a proper heirarchy the Inheritance process is going to work for us.



```python
class Grandparent:
    def __init__(self, nationality, baldness, anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def introduction(self):
        print(f"Hello my nationality is :{self.nationality} and about my anger_issue if you want to ask about then the answer is : {self.anger_issues}")


class Parent(Grandparent):
    def __init__(self, nationality, baldness, anger_issues, intelligence):
        super().__init__(nationality, baldness, anger_issues)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f"About my intelligence if you guys want to know that wheather if i am intelligent or not {self.intelligence}")


class Child(Parent):
    def __init__(self, nationality, baldness, anger_issues, intelligence, charm):
        super().__init__(nationality, baldness, anger_issues, intelligence)
        self.charm=charm

    def charm_details(self):
        print(f"About my charm if you guys want to know that wheather if i am charming or not then the {self.charm}")



parent_1=Parent("Indian", "Yes", "No", "Yes")
parent_1.introduction()
```

```python
child_1= Child("Indian", "No", "Yes", "Yes", "Yes")
child_1.introduction()
```

### 3. Multiple Inheritance
In multiple Inheritance, The properties of two parent class or multiple parent class will be getting inherited into the child class



```python
class Ironmans:
    def __init__(self, name, power, suit, weapon, nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"My name is {self.name} and powers are: {self.power}")


class Captainamerica:
    def __init__(self, name, power, suit, weapon, nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def weapon_details(self):
        print(f"I carry the weapon as : {self.weapon}")



class Hulk(Ironmans, Captainamerica):
    def __init__(self, name, power, suit, weapon, anger_issues, nationality="American"):
        super().__init__( name, power, suit, weapon, nationality)
        self.anger_issues=anger_issues

    def anger_issues(self):
        print(f"About my anger issues if i am having or not then the answer is: {self.anger_issues}")



hulk_guy=Hulk("Hulk", "Amazing Power and Strength", "Mark 85 Ironman Suit", "Repulsor beam and Shield" , "American", "Yes")
```
