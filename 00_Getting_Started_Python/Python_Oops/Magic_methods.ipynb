## Magic Methods
Magic methods are also called as Dunder methods because they have double  underscores __ before and after their name and they are special methods in python that allow us to customize how objects behave

### These Methods can help us define:
- 1. Object Initialization (__init__)
- 2. String Representation (__str__, __repr__)
- 3. Arithmetic Operation (__add__, __sub__, etc)
- 4. Comparison Operation (__eq__, __it__)
- 5. Length, Indexing, Iteration (__len__, getitem__,  etc)

### Why to use  Magic Methods

- 1. Makes objects behave like built in types (eg, list, integers)
- 2. Improves Redability by allowing  basic operation object1 + object2
- 3. Simplifies code by replacing verbose function calls with natural syntax.



### Python Dunder Methods

#### 1. `__init__()`

The `__init__()` method is a dunder method that is automatically called when an object is created.


```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Vinod", 27)

print(student.name)
print(student.age)
```

#### Output

```text
Vinod
27
```

When we write:

```python
student = Student("Vinod", 27)
```

Python automatically calls:

```python
__init__()
```


#### 2. `__str__()`


The `__str__()` method defines the human-readable string representation of an object.


```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old"


student = Student("Vinod", 27)

print(student)
```

#### Output

```text
Vinod is 27 years old
```

When we write:

```python
print(student)
```

Python internally calls:

```python
student.__str__()
```

#### 3. `__len__()`

The `__len__()` method defines what should happen when the `len()` function is used with an object.


```python
class Team:

    def __init__(self):
        self.members = ["A", "B", "C", "D"]

    def __len__(self):
        return len(self.members)


team = Team()

print(len(team))
```

#### Output

```text
4
```

When we write:

```python
len(team)
```

Python calls:

```python
team.__len__()
```


#### 4. `__add__()`

The `__add__()` method defines the behavior of the `+` operator for objects.


```python
class Product:

    def __init__(self, price):
        self.price = price

    def __add__(self, other):
        return self.price + other.price


product1 = Product(500)
product2 = Product(300)

print(product1 + product2)
```

#### Output

```text
800
```

When we write:

```python
product1 + product2
```

Python internally calls:

```python
product1.__add__(product2)
```

#### 5. `__eq__()`

The `__eq__()` method defines how two objects should be compared using the `==` operator.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        return self.name == other.name


student1 = Student("Vinod")
student2 = Student("Vinod")

print(student1 == student2)
```

#### Output

```text
True
```

When we write:

```python
student1 == student2
```

Python internally calls:

```python
student1.__eq__(student2)
```

#### Important Dunder Methods

| Dunder Method   | Used For                          |
| --------------- | --------------------------------- |
| `__init__()`    | Object initialization             |
| `__str__()`     | String representation of object   |
| `__len__()`     | `len()` function                  |
| `__add__()`     | `+` operator                      |
| `__sub__()`     | `-` operator                      |
| `__mul__()`     | `*` operator                      |
| `__eq__()`      | `==` operator                     |
| `__lt__()`      | `<` operator                      |
| `__gt__()`      | `>` operator                      |
| `__getitem__()` | `obj[index]`                      |
| `__setitem__()` | `obj[index] = value`              |
| `__call__()`    | Calling an object like a function |
