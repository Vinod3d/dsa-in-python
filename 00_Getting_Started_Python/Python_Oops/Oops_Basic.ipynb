## Object Oriented Programming



In Object Oriented Programming we are trying to match up wiht the real world scenario using programming

IN Oops we are going to learn about classes and objects

By a class we mean to provide a blueprint that must remain the same for each and eery single object for us

These Objects can have different functionalities from each other and different characteristics from each other as well but they must follow the similar blueprint

By a Blueprint or a class what we mean to have a means of providing the funtionalities and the characteristics to our objects

```python
class Avenger:
    def __init__(self, name, weapon, nationality="American"):
        self.name=name
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def carrying_weapon(self):
        print(f"I am carrying the weapoon {self.weapon}")
```

```python
Iron_man=Avenger("Iron man", "Genius intellect and Technology", "Repulsor beams")
Iron_man.introduction()
Iron_man.carrying_weapon()
```

```python
Captain_America=Avenger("Captain America", "Super strength and agility", "Shield")
Captain_America.introduction()
Captain_America.carrying_weapon()
```

```python
Thor=Avenger("Thor", "God of Thunder","Hammer", "Asgardian")
Thor.introduction()
Thor.carrying_weapon()
```

```python
class Cartoon:
    def __init__(self, name, show, catchphrase):
        self.name=name
        self.show=show
        self.catchphrase=catchphrase

    def introduction(self):
        print(f"Hellow my name is {self.name}")

    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
         
```

```python
Mickey_mouse=Cartoon("Mickey Mouse", "Mickey Mouse Clubhouse", "oh boy !")
Mickey_mouse.show
```
