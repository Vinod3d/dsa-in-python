## Object Oriented Programming



In Object Oriented Programming we are trying to match up wiht the real world scenario using programming

IN Oops we are going to learn about classes and objects

By a class we mean to provide a blueprint that must remain the same for each and eery single object for us

These Objects can have different functionalities from each other and different characteristics from each other as well but they must follow the similar blueprint

By a Blueprint or a class what we mean to have a means of providing the funtionalities and the characteristics to our objects

```python
class Avenger:
    def __init__(self, name, weapon, nationality="American"):
        self.name=name
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def carrying_weapon(self):
        print(f"I am carrying the weapoon {self.weapon}")
```

```python
Iron_man=Avenger("Iron man", "Genius intellect and Technology", "Repulsor beams")
Iron_man.introduction()
Iron_man.carrying_weapon()
```

```python
Captain_America=Avenger("Captain America", "Super strength and agility", "Shield")
Captain_America.introduction()
Captain_America.carrying_weapon()
```

```python
Thor=Avenger("Thor", "God of Thunder","Hammer", "Asgardian")
Thor.introduction()
Thor.carrying_weapon()
```

### Object Oriented Programming Terminologies

**Attributes** ---> Attributes stands for the charecteristics that a specific  object is going to have

**Methods** ---> Functionality that a specific  object is going to have

**self** ---> It refer to the current object of the class

**constructor** ---> A Constructor is being refered by __init__, it gets automatically called as soon as we are going to create an object


#### Types of Constructor
1. **Default Constructor**  : it does'nt have the parameter inside it and it is only having the self keyword inside it. it assign a default value to all the objects.



```python
class Dog:
    def __init__(self):
        self.name="Labrador"
        self.color="Brown"
```

2. **Parameterized Constructor** : it consist of parameter inside it so that every object can have a specific attribute associated to it

```python
class Dog:
    def __init__(self, breed, color):
        self.breed=breed
        self.color=color
```

#### Type of Methods

1. **instance method**: instance method operate on object Attributes. The instance method is going to use self keyword to access and modify object attributes.



```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def is_pass(self):
        if self.marks >= 40:
            return True
        return False


# Creating objects
student1 = Student("Rahul", 75)
student2 = Student("Amit", 35)

student1.display()
print(student1.is_pass())

student2.display()
print(student2.is_pass())
```


2. **class method**:  It is the method of the whole class or the whole blueprint which will remain the same for all of the objects then such kind of method is being reffered as a class method.

    - In class method, these methods work with the class level attributes.

    - They are going to use `cls` instead of self

    - They are called using the class name as well

    - For specific that the current method is a class method we have to use the abbreviation as `@classmethod`

```python
class Employee:
    company = "TechCorp"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company


# Create objects
emp1 = Employee("Rahul", 50000)
emp2 = Employee("Amit", 60000)

print(emp1.company)
print(emp2.company)

# Change company for all employees
Employee.change_company("Google")

print(emp1.company)
print(emp2.company)
```

3. **Static method**: Static method do not use self or cls.

    - They work like normal functions but belong to a class.
    
    - They are going to be designed by a abbreviation @staticmethod

    - static method can also be called by the classname also

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @staticmethod
    def is_valid_salary(salary):
        return salary >= 15000


emp1 = Employee("Rahul", 25000)

print(emp1.name)
print(emp1.salary)

print(Employee.is_valid_salary(25000))
print(Employee.is_valid_salary(10000))
```
