## Operator Overloading

Operator overloading is the process of defining the behavior of an operator for objects of a user-defined class using special (dunder) methods.

```python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)
```

```python
a = Number(10)
b = Number(5)

print((a + b).value)   # 15
print((a - b).value)   # 5
```
