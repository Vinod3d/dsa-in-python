## Polymorphism

Polymorphism means Many Forms in Greek.

Polymorphism is the ability of different objects to respond to the same method or operation in different ways.

### Types of Polymorphism

#### 1. Method Overriding (Runtime Polymorphism)
A Child class provides a specific implementation of a method that has already been defined into the parent class. This allow Dynamic Selection at Runtime.

```python
class Grandparent:
    def speak(self):
        print("Hello My name is Amit")


class Parent(Grandparent):
    def speak(self):
        print("Hello My name is Vinod")

class Child(Parent):
    def speak(self):
        print("Hello My name is Vimal")
```

```python
def attribute_provider(object):
    print(object.speak())
```

```python
grandpa = Grandparent()
parent = Parent()
child = Child()
```

```python
attribute_provider(grandpa)
attribute_provider(parent)
attribute_provider(child)
```

#### 2. Method Overloading
Method overloading means having multiple methods with the same name but different parameters in the same class.

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

#### 3. Duck Typing
Duck typing means that an object's behavior is determined by the methods and properties it provides, rather than by its class or type.

```python
class Dog:
    def speak(self):
        print("Bark")


class Cat:
    def speak(self):
        print("Meow")


def make_sound(animal):
    animal.speak()


make_sound(Dog())
make_sound(Cat())
```
