## Streamlit

Streamlit is an open source Python Library used to build Beautiful, interactive web applications for data science and machine learning

### Why to use Streamlit

##### 1. Easy to use
##### 2. Fast Development
##### 3. Interactive UI Elements ---> Buttons, Sidebar, Text Input, Chart, etc
##### 4. Great for data science ---> Supports Pandas, Matplotlib, Plotly and other Libraries
##### 5. Deployment is easy
