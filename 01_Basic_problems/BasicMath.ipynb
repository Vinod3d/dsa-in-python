### 1. Sum of N Even Natural Numbers

```python
def sum_of_even_numbers(n):
    sum = 0
    for i in range(n*2+1):
        if i%2 == 0:
            sum+= i
    return sum
    
sum_of_even_numbers(3)
```

### 2. Check for Even Number

```python
def is_even(n):
    return True if n%2 == 0 else False

is_even(7)
```

### 3. Check for Prime Number

```python
import math

def is_prime(n):
    sqrt = int(math.sqrt(n)) 
    for i in range(2, sqrt+1):
        return True if n%i != 0 else False

is_prime(7)
        
    
```

### 4. Valid Perfect Square


```python
def is_perfect_square(num):
    low = 1
    high = num // 2
    while low<= high:
        mid = (low + high) // 2
        square = mid * mid
        if square == num:
            return True
        elif square < num:
            low = mid + 1
        else:
            high = mid - 1
    return False

is_perfect_square(16)
        
```

### 5. Decimal to Binary
Input: n = 5

Output: "101"

```python
def int_to_binary(n):
    if n == 0:
        return "0"

    is_negative = n < 0
    n = abs(n)

    binary = ""

    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    if is_negative:
        return "-" + binary

    return binary

int_to_binary(-255)
```

### 6. Binary to Decimal

```python
def binary_to_decimal(binary_str):
    is_negative = binary_str[0] == '-'
    if is_negative:
        binary_str = binary_str[1:]
    decimal = 0
    power = 0
    for i in range(len(binary_str)-1, -1, -1):
        decimal += int(binary_str[i]) * (2**power)
        power += 1
    if is_negative:
        return -decimal
    return decimal

binary_to_decimal("1010")
        
        
        
        
    
```

### 7. GCD of Two Number

“I use Euclidean algorithm which reduces time complexity to O(log n) instead of checking all divisors.”

```python
def gcd(n, m):
    while m != 0:
        n, m = m, n % m
    return n
```
