### 1. Reverse String

```python
def reverse_string(s):
    rev = ""
    for i in range(len(s)-1, -1, -1):
        rev+=s[i]
    return rev

reverse_string("Hello")
```

### 2. Count Vowels in a string

```python
def count_vowels(s):
    vowel = "aeiou"
    s = s.lower()
    count = 0
    for i in range(len(s)):
        if s[i] in vowel:
            count+=1
    return count

count_vowels("Hello, World")
        
```

### 3. Check for same strings

```python
def are_equal_strings(s, t):
    s_len = len(s)
    t_len = len(t)
    
    if s_len != t_len:
        return False

    for i in range(s_len):
        if s[i] != t[i]:
            return False

    return True

are_equal_strings("hello", "world")
    
```

### 4. Check Palindrome
A string is considered a palindrome if it reads the same forward and backward, ignoring spaces, punctuation, and case.

```python
def is_palindrome(s):
    cleaned = ""

    for ch in s:
        # check A-Z
        if 'A' <= ch <= 'Z':
            cleaned += chr(ord(ch) + 32)  # convert to lowercase

        # check a-z or 0-9
        elif ('a' <= ch <= 'z') or ('0' <= ch <= '9'):
            cleaned += ch

    # reverse manually
    reversed_str = ""
    for i in range(len(cleaned) - 1, -1, -1):
        reversed_str += cleaned[i]

    return cleaned == reversed_str


# Example
s = "A man, a plan, a canal: Panama"
print(is_palindrome(s))
```

### 5. Count words in a string

```python
def count_words(s):
    words = []
    count = 0
    curr_word = ""
    for ch in s:
        if ch != " ":
            curr_word += ch
        else:
            if curr_word != "":
                words.append(curr_word)
                count += 1
                curr_word = ""
    if curr_word != "":
        words.append(curr_word)
        count += 1
    return count

s = "hello bhai kaise ho"
result = count_words(s)
print(result)

                

    
```

### 6. Remove Duplicates in a string

```python

def remove_duplicates(s):
    mod = ""
    for ch in s:
        if ch not in mod:
            mod += ch
    return mod

remove_duplicates("Hello, World!")
```

### 7. Count consonants in a string

```python
def count_consonants(s):
    count = 0
    vowel = "aeiou"
    for ch in s:
        if 'A' <= ch <= 'Z':
            ch = chr(ord(ch) + 32)
        if ch not in vowel and 'a' <= ch <= 'z':
            count += 1
    return count

count_consonants("Hello, World!")
```

### 8. Check for anagrams
An anagram is a word or phrase formed by rearranging the characters of a different word or phrase, using all the original characters exactly once.

```python
def is_anagram(s, t):
    if len(s) != len(t):
        return False

    count = {}

     # count characters of s
    for ch in s:
       if ch in count:
           count[ch] += 1
       else:
           count[ch] = 1
           
    # subtract using t
    for ch in t:
        if ch not in count:
            return False

        count[ch] -=1

        if count[ch] < 0:
            return False
    return True

        
        

is_anagram('anagram', "nagaram")
```

### 9. Check Subsequence
You are given two strings `s` and `t`. Your task is to determine if string t is a subsequence of string s. A subsequence of a string is a new string that is formed from the original string by deleting some (or no) characters without changing the order of the remaining characters.



```python
def is_subsequence(s, t):
    i = 0  # pointer for t

    for ch in s:
        if i < len(t) and ch == t[i]:
            i += 1

    return i == len(t)


# Test
print(is_subsequence("abcde", "ace"))
```

### 10. Check for Substring
You are given two strings, s and t. Your task is to determine if the string t is a substring of the string s. A substring is a contiguous sequence of characters within a string. Do not use any built-in functions for string operations and do not use recursion.

```python
def is_substring(s, t):
    t_len = len(t)
    s_len = len(s)
    if t_len == 0:
        return True
    for i in range(s_len - t_len + 1):
        if s[i] == t[0]:
            if s[i: i + t_len] == t:
                return True
    return False
    
is_substring(s = "hello world", t = "world")
    
```

### 11. Length of the Longest Word

```python
def longest_word_length(s):
    word = ""
    w_len = 0
    for ch in s:
        if ch != " ":
            word += ch
        else:
            if word != "":
                if w_len <= len(word):
                    w_len = len(word)
                word = ""
    # last word ke liye
    if word != "":
        if w_len < len(word):
            w_len = len(word)

    return w_len
            
            
            


longest_word_length(s = "The quick brown fox jumps over the lazy dog")
```
