## Array in Python

An array is a data structure that stores multiple elements of the same data type in a fixed-size sequence of memory locations. Each element is accessed using an index (position), which usually starts from 0.

Arrays are commonly used when you know how many elements you need in advance.

```
int numbers[3] = {10, 20, 30};
```

#### Key Features of Array
- Stores elements of same data type
- Fixed size (cannot easily change after creation)
- Fast access using index
- Stored in contiguous memory

````python
import array

arr = array.array('i', [1, 2, 3, 4])  # 'i' → integer type
print(arr)
````

### List in Python
A list is a flexible data structure that can store multiple elements of different data types, and its size can grow or shrink dynamically.

Lists are commonly used in modern programming languages like Python, Java, and C#.

#### Example:

```python
List = [10, "Apple", 3.14, True]

```

#### Key Features of List
- Dynamic size (can grow or shrink)
- Can store different data types
- Easy to add or remove elements
- Supports many built-in operations

````python
lst = [1, "hello", 3.5, True]
print(lst)
````

# Types of Searching Techniqes

## Linear Search 

Linear Search is a searching algorithm in which each element in a sequence is compared with the target value one by one until the target is found or the entire sequence is checked.

````python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [4, 9, 2, 7, 5]
target = 7

linear_search(arr, target)
````

## Binary Search
Binary Search is an efficient searching algorithm that works on sorted data only.

Instead of checking every element (like Linear Search), it:

- Divides the list into half
- Checks the middle element
- Decides which half to continue searching

````python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif target < arr[mid]:
            right = mid - 1
        else:
            left = mid + 1

    return -1

arr = [2, 5, 8, 12, 16]
target = 12
binary_search(arr, target)
````

# Types of Sorting Techniques

Sorting algorithms are mainly classified into two categories:

### (a) Comparison-Based Sorting

These algorithms compare elements to sort them.

* Bubble Sort
* Selection Sort
* Insertion Sort
* Merge Sort
* Quick Sort
* Heap Sort

### (b) Non-Comparison-Based Sorting

These algorithms do not compare elements directly.

* Counting Sort
* Radix Sort
* Bucket Sort

---




## 1. Bubble Sort
Bubble Sort is a simple sorting algorithm that repeatedly compares adjacent elements and swaps them if they are in the wrong order.

````python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                # swap
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr

arr = [5, 3, 8, 4]
bubble_sort(arr)
````

## 2. Selection Sort
Selection Sort is a sorting algorithm where:

- We repeatedly select the smallest element from the unsorted part
- And place it at the correct position (beginning)

````python
def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # swap
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr

arr = [64, 25, 12, 22, 11]

selection_sort(arr)
````

## 3. Insertion Sort
Insertion Sort is a simple sorting algorithm that builds the final sorted list one element at a time by taking each element and inserting it into its correct position within the already sorted part of the list.

````python
def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        key = arr[i]  
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr
arr = [5, 3, 4, 1]
insertion_sort(arr)
````
